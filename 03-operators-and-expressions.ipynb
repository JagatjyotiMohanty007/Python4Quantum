## 3.1 Arithmetic operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+ - *` | add, subtract, multiply | `7 * 2` | `14` |
| `/` | true division (always `float`) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | exponent | `7 ** 2` | `49` |

```
a, b = 7, 2
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)
print(divmod(7, 2))                 # (3, 1) -- quotient and remainder together

# Modular arithmetic: wrap-around and cycles
print((7 ** 4) % 15)                # 1
print(pow(7, 4, 15))                # same result, faster for huge numbers
print((10 + 5) % 12)                # clock arithmetic -> 3

# Floor division and modulo map a flat index onto a grid
flat_index, row_length = 7, 4
row, col = flat_index // row_length, flat_index % row_length
print(f"index {flat_index} -> row {row}, col {col}")   # row 1, col 3

# Normalising with a square root
import math
n = 3
print(1.0 / math.sqrt(2 ** n))
```

> Careful with negatives: `-7 // 2` is `-4` (floor, not truncation) and `-7 % 2` is `1`.

## 3.2 Comparison operators

`==  !=  <  <=  >  >=` return booleans. Python allows **chained comparisons**:

```
x = 5
print(x == 5, x != 5, x < 10, x >= 4)

angle = 4.71
print(0.0 <= angle < 6.2832)        # chained: 0 <= angle and angle < 6.2832

# Never compare floats with == after arithmetic
import math
print(0.1 + 0.2 == 0.3)                       # False
print(math.isclose(0.1 + 0.2, 0.3))           # True
tolerance = 0.99
score = 0.995
print(score >= tolerance)
```

## 3.3 Logical operators and short-circuiting

`and`, `or`, `not` combine conditions. They **short-circuit**:

- `a and b` evaluates `b` only if `a` is truthy.
- `a or b` evaluates `b` only if `a` is falsy.

They return one of the *operands*, not necessarily `True`/`False`.

```
ready = True
server_ok = False
print(ready and not server_ok)     # True
print(ready or server_ok)          # True

print(False and (1 / 0))           # False -- the division never runs

# Guard expression: validate in a safe order
requested = 1024
available = True
limit = 4096
can_run = available and (requested > 0) and (requested <= limit)
print(can_run)

# Short-circuit prevents errors on empty data
samples = {}
valid = (len(samples) > 0) and (sum(samples.values()) > 0)
print(valid)

# `or` supplies a default for falsy values
user_name = ""
print(user_name or "anonymous")    # anonymous
```

## 3.4 Bitwise operators

Bitwise operators work on the binary digits of integers.

| Operator | Name | Example (`a=0b1010`, `b=0b1100`) |
|---|---|---|
| `&` | AND | `0b1000` |
| `\|` | OR | `0b1110` |
| `^` | XOR | `0b0110` |
| `~` | NOT | `-11` |
| `<<` | left shift | `a << 1 = 0b10100` |
| `>>` | right shift | `a >> 1 = 0b0101` |

```
a = 0b1010   # 10
b = 0b1100   # 12
print(bin(a & b), bin(a | b), bin(a ^ b), ~a, bin(a << 1), bin(a >> 1))

# XOR truth table: output is 1 when inputs differ
print("x y  x^y")
for x in (0, 1):
    for y in (0, 1):
        print(f"{x} {y}   {x ^ y}")

# Extracting one bit with shift + mask
value = 0b101101
position = 3
print((value >> position) & 1)     # 1

# Flags: set, clear, toggle, test
READ, WRITE, EXEC = 0b001, 0b010, 0b100
perms = READ | WRITE               # set two flags
print(bin(perms), bool(perms & WRITE), bool(perms & EXEC))
perms &= ~WRITE                    # clear WRITE
perms ^= EXEC                      # toggle EXEC
print(bin(perms))
```

A useful identity: `x ^ x == 0` and `x ^ 0 == x`, so XOR toggles bits and is its own inverse.

## 3.5 Assignment operators

Compound assignment updates a variable in place: `+= -= *= /= //= %= **= &= |= ^= <<= >>=`.

```
total = 10
total += 5
total *= 2
print(total)           # 30

mask = 0b0001
mask |= 0b0010         # set bit 1
mask ^= 0b0001         # flip bit 0
print(bin(mask))       # 0b10
```

## 3.6 The walrus operator `:=`

Introduced in Python 3.8 (PEP 572), `:=` assigns **as part of an expression**. It avoids computing a value twice.

```
if (n := len([1, 2, 3])) > 2:
    print(n)                             # 3

readings = ["ok", "err", "ok", "ok", "warn", "ok"]
if (ok_count := readings.count("ok")) > 3:
    print(f"'ok' appeared {ok_count} times")

# Typical use: loop until a condition fails
data = iter([4, 8, 15, 0, 16])
while (value := next(data)) != 0:
    print("processing", value)
```

Use it sparingly — clarity first.

## 3.7 Membership and identity operators

- `in`, `not in` — is a value inside a collection (or substring in a string)?
- `is`, `is not` — are two names bound to the *same object*?

```
ids = [0, 1, 2]
print(1 in ids, 5 not in ids)
print("th" in "python")

a = [1, 2]
b = [1, 2]
c = a
print(a == b, a is b, a is c)    # True False True
print(None is None)

value = None
if value is None:                # correct way to test for None
    print("no value")
```

`==` asks "equal values?", `is` asks "same object?". Use `is` only for `None`, `True`, and `False`.

## 3.8 Precedence, evaluation order, and parentheses

From highest to lowest (simplified): `**` → unary `-x ~x` → `* / // %` → `+ -` → shifts → `&` → `^` → `|` → comparisons / `in` / `is` → `not` → `and` → `or`.

```
print(2 + 3 * 4)        # 14
print((2 + 3) * 4)      # 20 -- parentheses override precedence
print(2 ** 3 ** 2)      # 512 -- ** groups right to left: 2 ** (3 ** 2)
print(-2 ** 2)          # -4  -- ** binds tighter than unary minus
print(10 - 4 - 3)       # 3   -- most operators group left to right
print(not True or True) # True -- (not True) or True
```

When in doubt, add parentheses: they cost nothing and make intent obvious.

## Common pitfalls

- `==` between floats; use `math.isclose`.
- Using `is` to compare numbers or strings.
- `a & b` (bitwise) versus `a and b` (logical) — they are not interchangeable.
- Relying on precedence you have to look up.
