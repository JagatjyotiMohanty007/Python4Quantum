## 2.1 Tokens

A **token** is the smallest unit the interpreter recognises. Every program is a stream of tokens in five categories:

| Category | Examples |
|---|---|
| Keywords | `def`, `class`, `import`, `return`, `with`, `as` |
| Identifiers | `total`, `calculate_tax`, `Invoice` |
| Literals | `42`, `3.14159`, `"text"`, `True`, `None` |
| Operators | `+`, `*`, `^`, `&` |
| Delimiters | `( )`, `[ ]`, `{ }`, `:`, `,` |

## 2.2 Keywords and the `keyword` module

Keywords are reserved words with special meaning. You cannot use them as names. The `keyword` module lists them for your interpreter version.

```
import keyword

print(len(keyword.kwlist))        # number of keywords (varies by version)
print(keyword.kwlist[:10])
print(keyword.iskeyword("class"))     # True
print(keyword.iskeyword("my_var"))    # False
print("if" in keyword.kwlist, "widget" in keyword.kwlist)
```

## 2.3 Identifiers and naming conventions

An **identifier** is a name for a variable, function, class, or module. Rules:

- Start with a letter or underscore, never a digit
- Contain only letters, digits, and underscores (Unicode letters are allowed in Python 3)
- Are case-sensitive: `Total` and `total` are different
- Cannot be a keyword

Conventions from PEP 8:

| Kind | Style | Example |
|---|---|---|
| Variables, functions | `snake_case` | `total_marks`, `calculate_total` |
| Classes | `PascalCase` | `BankAccount` |
| Constants | `ALL_CAPS` | `MAX_RETRIES` |
| "Private" by convention | leading underscore | `_cache` |

```
name = "Alice"
age = 25
total_marks = 450
calculate_total = 100 + 50
_internal_state = 1.0
alpha_value = 0.7071
α_param = 0.7071            # legal (Unicode identifier), but avoid in shared code

class BankAccount:           # PascalCase for classes
    pass

print(name, age, total_marks, calculate_total, alpha_value == α_param)
```

## 2.4 Literals

A **literal** is a fixed value written directly in code.

```
title = "Python"       # string literal
count = 20             # integer literal
big = 1_000_000        # underscores improve readability
mask = 0b1010          # binary literal (10)
perm = 0o755           # octal literal (493)
colour = 0xFF          # hexadecimal literal (255)
pi_approx = 3.14       # float literal
z = 0.5 + 0.5j         # complex literal
is_ready = True        # boolean literal
nothing = None         # the None literal
items = [1, 2, 3]      # list literal
print(title, count, big, mask, perm, colour, pi_approx, z, is_ready, nothing, items)
```

## 2.5 Constants

Python does not enforce constants. By convention, `ALL_CAPS` names are treated as constants and should not be reassigned.

```
MAX_CONNECTIONS = 8192
TAX_RATE = 0.18
DEFAULT_TIMEOUT = 30
print(MAX_CONNECTIONS, TAX_RATE, DEFAULT_TIMEOUT)
```

## 2.6 Variables and name binding

A variable is **a name bound to an object**, not a labelled box of memory. Assignment creates or changes the binding; the type belongs to the *object*, not the name.

```
student_name = "Sam"
student_age = 18
has_paid = True
items_sold = 1024
items_sold = items_sold * 2     # rebinding: the name now points at a new int
print(student_name, student_age, has_paid, items_sold)

a = 5
a = a + 1                       # evaluate right side first, then rebind
print(a)                        # 6
```

Because types live on objects, the same name can be rebound to a different type (*dynamic typing*) — legal, but usually confusing:

```
x = 1
print(type(x))      # <class 'int'>
x = "one"
print(type(x))      # <class 'str'>
```

Python infers types at runtime; you never declare them. (Optional *type hints* are introduced in Module 15.)

## 2.7 Built-in types at a glance

| Type | Meaning | Example |
|---|---|---|
| `int` | whole number, arbitrary size | `42` |
| `float` | 64-bit decimal (IEEE 754) | `3.14` |
| `complex` | complex number | `2+3j` |
| `str` | immutable text | `"hello"` |
| `bool` | `True` / `False` | `True` |
| `NoneType` | "no value" | `None` |
| `list`, `tuple`, `dict`, `set` | collections (Module 6) | `[1, 2]` |

```
n = 7
price = 3.5
flag = True
nothing = None
print(type(n), type(price), type(flag), type(nothing))

print(2 ** 100)          # ints never overflow
print(0.1 + 0.2)         # 0.30000000000000004 -- floats are binary approximations
print(0.1 + 0.2 == 0.3)  # False! Compare floats with math.isclose (Module 3)

message = "Hello, world!"
print(message.upper(), len(message))

result = None
print(result is None)    # use `is None`, not `== None`
```

## 2.8 Type conversion

Use the type name as a function: `int()`, `float()`, `str()`, `bool()`. Conversion can fail with `ValueError`.

```
s = "42"
print(int(s), float(s), str(42))
print(int(3.9))            # 3 -- truncates toward zero, does not round
print(round(3.9))          # 4

# Number bases
bits = "1101"
value = int(bits, 2)               # parse text as base 2
print(value)                       # 13
print(f"{value:04b}")              # 1101 -- back to 4-digit binary
print(bin(value), hex(value), oct(value))   # 0b1101 0xd 0o15

print(float("3.14159265"))
# int("abc")  # would raise ValueError: invalid literal for int()
```

## 2.9 Truthiness

Every object has a truth value. In an `if` or `while`, Python uses it automatically.

**Falsy values:** `False`, `None`, `0`, `0.0`, `0j`, `""`, `[]`, `()`, `{}`, `set()`. Almost everything else is truthy.

```
print(bool(""), bool("q"), bool(0), bool(1))

values = [0, "", [], {}, None, False]
print([bool(v) for v in values])         # all False

items_left = 0
print(bool(items_left))                  # False -> "nothing left"

basket = []
if not basket:                           # idiomatic emptiness test
    print("basket is empty")
```

## Common pitfalls

- Comparing floats with `==`.
- `int("3.5")` fails — convert via `int(float("3.5"))`.
- Using `== None` instead of `is None`.
- Shadowing built-ins by naming a variable `list`, `str`, or `sum`.
