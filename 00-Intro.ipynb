# Python Programming: From First Program to Professional Practice

A single, sequenced course that takes a learner from "What is a variable?" to writing tested, documented, well-structured Python applications. It merges two source tutorials — a basic one and an intermediate-to-advanced one — into one continuous path, with **generic Python examples throughout**.

## Who this is for

- **Absolute beginners** who have never programmed
- **Learners moving up** from basic syntax to idiomatic, professional Python
- **Instructors** who want a ready-made syllabus with runnable examples and practice tasks

## Requirements

- **Python 3.10 or newer** (Module 4 uses `match`/`case`; a few later modules use 3.9+ and 3.10+ typing syntax)
- A code editor such as VS Code, or Jupyter
- No internet connection is needed after installation; every example runs offline using only the standard library

## Module format

Every module follows the same shape:

1. **What you'll learn** — the goals
2. Numbered sections, each with an explanation and **runnable code**
3. **Common pitfalls** — mistakes that catch learners out
4. **Practice** — three short exercises

## Suggested pacing

| Audience | Plan |
|---|---|
| Complete beginner | one module per session, in order; do all practice tasks |
| Has programmed before | skim Part I, start properly at Module 6 |
| Intermediate | start at Module 10 or 11; use Part IV and the capstone as the goal |

A typical full course is roughly 40–60 hours including practice and the capstone.

## How to use the code

- Run snippets in a notebook cell, the Python REPL, or save them as `.py` files.
- Snippets within a module build on each other, so run them in order.
- Lines such as `# int("abc")  # would raise ValueError` are deliberately commented out; uncomment one to see the error, then comment it again.
- Examples that touch the disk use temporary folders, so nothing is left behind.
