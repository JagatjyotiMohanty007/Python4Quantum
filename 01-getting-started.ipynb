## 1.1 Checking your environment

Before writing real programs, confirm Python is installed and see which version you are running. Different versions behave slightly differently (for example, `match`/`case` needs Python 3.10+), so this is a habit worth building early.

```
import sys

print("Python version:", sys.version)
print("Environment is working correctly!")
```

> **Course requirement:** Python 3.10 or newer. A few advanced modules use 3.9+/3.10+ syntax and say so where it matters.

## 1.2 Comments

Comments are ignored by Python and explain code to humans. `#` starts a comment that runs to the end of the line.

```
# this is a single-line comment
x = 10
print(x)

y = 20  # an inline comment
print(y)
```

Python has no dedicated multi-line comment syntax. Two common approaches:

1. Several lines that each start with `#`.
2. A triple-quoted string. When it is the first statement in a function, class, or module it becomes a **docstring** — documentation that tools and `help()` can read.

```
def divide_numbers(dividend, divisor):
    """Divide two numbers.

    Args:
        dividend (float): The number to be divided.
        divisor (float): The number to divide by.

    Returns:
        float: The result of the division.

    Raises:
        ValueError: If the divisor is zero.
    """
    if divisor == 0:
        raise ValueError("Cannot divide by zero.")
    return dividend / divisor


print(divide_numbers(10, 2))      # 5.0
print(divide_numbers.__doc__[:20])  # docstrings are stored on the function
```

> A stray triple-quoted string in the middle of code is just an unused string value, not a true comment. Use `#` for notes and triple quotes for docstrings.

## 1.3 Output — `print()`

`print()` is how a program talks to you. You write code, you see a result — that feedback loop is what makes programming feel real.

```
print("Hello, World!")
print("Python is fun to learn.")
print("Numbers work too:", 42)
```

Two keyword arguments control formatting:

- `sep` — the separator placed between arguments (default: a space)
- `end` — what is printed after the last argument (default: a newline)

```
print("a", "b", "c", sep=" | ")   # a | b | c
print("loading", end="...")
print("done")                     # loading...done
print("one line", end="!\n")
```

## 1.4 Input — `input()`

`input()` pauses the program, shows a prompt, and returns whatever the user types **as a string** — even if it looks like a number.

```
name = input("What is your name? ")
print("Hello,", name, "- welcome to Python!")

raw = input("How many items? ")   # the user types 12
count = int(raw)                  # convert text to a number
print(count + 1, type(count))
```

Forgetting the conversion is the classic beginner bug: `"12" + 1` raises a `TypeError`, while `int("12") + 1` gives `13`. Conversions are covered in Module 2.

## 1.5 Statements and expressions

- A **statement** is a complete instruction: an assignment, an import, a loop, a function definition, and so on.
- An **expression** is anything that *evaluates to a value*. Statements use expressions to do their work.

```
# Statements
x = 10                         # assignment statement
import math                    # import statement
if x > 5:                      # conditional statement
    print("x is greater than 5")
for i in range(3):             # loop statement
    print(i)

def square(n):                 # function definition statement
    return n * n               # return statement

# Expressions: each one produces a value
42                             # literal
3 + 5                          # arithmetic
x > 5                          # comparison -> True
len("hello")                   # function call -> 5
[1, 2, 3][0]                   # indexing -> 1
"even" if x % 2 == 0 else "odd"   # conditional expression
"q" in "python"                # membership -> False
print(square(4), math.sqrt(16))
```

A handy test: if you can put it on the right-hand side of `=` or inside `print(...)`, it is an expression.

## Common pitfalls

- Mixing tabs and spaces for indentation (use 4 spaces).
- Forgetting that `input()` returns text.
- Treating `"""..."""` in the middle of code as a comment instead of a string.
