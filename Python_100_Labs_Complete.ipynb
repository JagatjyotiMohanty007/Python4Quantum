## Comments
Comments are ignored by Python and help explain code to humans. `#` starts a single-line comment.

````python
# this is a single-line comment
x = 10
print(x)
y = 20  # Inline comment
print(y)
"""
This is a multi-line comment style in Python.
It is usually used for long explanations or documentation.
"""
def divide_numbers(dividend, divisor):
    """Divide two numbers.
    This function performs float division and raises an error
    if the divisor is set to zero.
    Args:
        dividend (float): The number to be divided.
        divisor (float): The number to divide by.
    Returns:
        float: The result of the division.
    """
    if divisor == 0:
        raise ValueError("Cannot divide by zero.")
    return dividend / divisor
# No additional return statement is needed.
print(divide_numbers(10, 2))
````

## Recursion
A recursive function calls itself to solve a smaller version of the same problem.

````python
def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n - 1)
countdown(3)
````

## Base case
Every recursive function needs a base case to stop calling itself.

````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
````

## Nonlocal variables
`nonlocal` lets an inner function modify a variable in an enclosing function.

````python
def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
        return count
    return inner
inc = outer()
print(inc(), inc())
````

````python
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply
triple = make_multiplier(3)
print(triple(4))
````

## Decorators
A decorator wraps a function to change or extend its behavior.

````python
def log_calls(func):
    def wrapper(*args, **kwargs):
        print('calling', func.__name__)
        return func(*args, **kwargs)
    return wrapper
@log_calls
def add(a, b):
    return a + b
print(add(2, 3))
````

## functools.wraps
`wraps` preserves metadata from the original function when wrapping it.

````python
import functools
def deco(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
@deco
def hello():
    return 'hi'
print(hello.__name__)
````

## __enter__ and __exit__
Context managers implement `__enter__` and `__exit__` to define setup/teardown behavior.

````python
class Session:
    def __enter__(self):
        print('enter')
        return self
    def __exit__(self, exc_type, exc, tb):
        print('exit')
with Session():
    print('inside')
````

## Iterator protocol
The iterator protocol is implemented with `__iter__` and `__next__`.

````python
class Counter:
    def __iter__(self):
        self.i = 0
        return self
    def __next__(self):
        if self.i >= 3:
            raise StopIteration
        self.i += 1
        return self.i
print(list(Counter()))
````

## Memoization
Memoization caches results to avoid repeat computation.

````python
cache = {}
def fib(n):
    if n in cache:
        return cache[n]
    if n < 2:
        return n
    cache[n] = fib(n - 1) + fib(n - 2)
    return cache[n]
print(fib(10))
````
