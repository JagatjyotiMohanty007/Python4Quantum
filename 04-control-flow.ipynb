## 4.1 Blocks, suites, and indentation

Python uses **indentation**, not braces, to define blocks. A **suite** is the indented block attached to a compound statement (`if`, `for`, `def`, `class`, …). Use 4 spaces per level, consistently.

```
if True:
    print("inside the block")
print("outside the block")

def greet():
    print("hello")      # the suite of greet()

greet()
```

An `IndentationError` means your blocks are inconsistent — often a tab/space mix.

## 4.2 `if`, `else`, and `elif`

```
items = 512

if items < 1024:
    print("low item count")

if items < 1024:
    print("low")
else:
    print("high")

# Conditions are tested top to bottom; the first match wins
items = 1024
if items < 1024:
    print("low")
elif items == 1024:
    print("default")
else:
    print("high")
```

Because Python evaluates `elif` branches in order, put the most specific (or cheapest) tests first. A realistic example — choosing a setting from a measured value:

```
error_rate = 0.015

if error_rate < 0.001:
    level = 1      # near-ideal: fast path
elif error_rate < 0.01:
    level = 2      # standard
else:
    level = 3      # heavy processing

print(f"Selected level: {level}")
```

## 4.3 Nested conditionals

You can place an `if` inside another. Prefer combining conditions with `and` when the nesting adds nothing.

```
x = 8
if x > 0:
    if x % 2 == 0:
        print("positive even")

# Flatter equivalent
if x > 0 and x % 2 == 0:
    print("positive even")
```

Deep nesting is a code smell. **Guard clauses** (return early on bad input) keep functions flat — see Module 8.

## 4.4 The ternary (conditional) expression

`value_if_true if condition else value_if_false` is an *expression*, so it can sit inside assignments, f-strings, and arguments.

```
items = 512
status = "low" if items < 1024 else "high"
print(status)

x = 2
print("even" if x % 2 == 0 else "odd")

bit = 1
label = "|1>" if bit == 1 else "|0>"
print(label)

inverted = True
delta = -5 if inverted else 5
print(delta)
```

Do not chain ternaries more than once; use `if`/`elif` instead.

## 4.5 Structural pattern matching: `match` / `case`

`match` compares a value against **patterns** and runs the first one that fits. It is more than a switch: patterns can *destructure* sequences and mappings and add `if` guards.

### Literal and or-patterns

```
command = "stop"

match command:
    case "start" | "go":
        print("starting")
    case "stop":
        print("stopping")
    case _:                        # wildcard: matches anything
        print("unknown command")
```

### Sequence patterns with capture and guards

```
def describe(instruction: tuple) -> str:
    """Interpret a (verb, *arguments) tuple."""
    match instruction:
        case ("noop",):
            return "do nothing"
        case ("move", x):
            return f"move by {x}"
        case ("jump" | "swap" as verb, a, b) if a != b:
            return f"{verb} between {a} and {b}"
        case ("jump" | "swap", a, b):
            return "source and target must differ"
        case _:
            return "unrecognised instruction"


print(describe(("noop",)))
print(describe(("move", 3)))
print(describe(("swap", 0, 1)))
print(describe(("swap", 2, 2)))
print(describe(("teleport", 9)))
```

### Mapping patterns

```
def handle(event: dict) -> str:
    match event:
        case {"type": "click", "x": x, "y": y}:
            return f"click at ({x}, {y})"
        case {"type": "key", "key": k}:
            return f"key {k}"
        case _:
            return "ignored"


print(handle({"type": "click", "x": 10, "y": 20}))
print(handle({"type": "key", "key": "Enter"}))
print(handle({"type": "scroll"}))
```

`match` requires Python 3.10+. On older versions, use `if`/`elif`.

## Common pitfalls

- Using `=` instead of `==` in a condition (a `SyntaxError`, which is the safe outcome).
- Forgetting the colon after `if`, `elif`, `else`, or `case`.
- In `match`, a bare name like `case x:` is a **capture**, not a comparison against a variable. To match constants, use literals or dotted names.
- Ordering `elif` branches so an earlier, broader test hides a later one.
