## 7.1 Strings are immutable sequences

A string is an ordered, **immutable** sequence of characters. You can index and slice it like a list, but you cannot change a character in place.

```
word = "quantum"
print(word[0], word[-1], word[1:4], word[::-1])
print(len(word))

# word[0] = "Q"   # TypeError: 'str' object does not support item assignment
word = "Q" + word[1:]      # build a new string instead
print(word)
```

Multi-line strings use triple quotes; raw strings (`r"..."`) leave backslashes alone, which matters for paths and regular expressions.

```
text = """This is a
multi-line string."""
print(text)

print(r"C:\new\table")      # backslashes preserved
print("tab:\tend")           # escape sequences are interpreted
```

## 7.2 Everyday string methods

```
label = "  q0_measured  "
print(label.strip())                       # remove surrounding whitespace
print(label.strip().replace("measured", "final"))

word = "Qubit"
print(word.lower(), word.upper(), word.title(), word.swapcase())

csv_line = "ada,grace,linus"
print(csv_line.split(","))                 # ['ada', 'grace', 'linus']
print("a b  c".split())                    # splits on any whitespace
print(" | ".join(["x0", "x1", "x2"]))      # join is called on the separator

print("python".startswith("py"), "python".endswith("on"))
print("banana".count("an"), "banana".find("nan"), "banana".find("zzz"))
print("42".isdigit(), "abc".isalpha(), "   ".isspace())
print("a-b-c".rsplit("-", 1))
print("hello".center(11, "*"))
print("7".zfill(3))
```

Remember that string methods **return new strings**; the original is unchanged. That makes **method chaining** natural:

```
print("  q0  ".strip().upper())
raw = "  Hello, World!  "
print(raw.strip().lower().replace(",", "").split())
```

## 7.3 String formatting

Three styles exist; prefer **f-strings** (Python 3.6+).

```
shots = 1024
name = "report"

print("ran " + str(shots) + " shots")        # concatenation (clumsy)
print("ran {} shots".format(shots))           # str.format
print(f"ran {shots} shots")                   # f-string: best
print(f"hello {name.upper()}")                # any expression works
print(f"{shots * 2=}")                        # debugging form: shows expression and value
```

Format specifiers go after a colon:

```
pi = 3.14159265
n = 13
big = 1234567.891

print(f"{pi:.2f}")          # 3.14   -- 2 decimal places
print(f"{n:04d}")           # 0013   -- zero-pad to width 4
print(f"{n:08b}")           # 00001101 -- binary
print(f"{n:x} {n:X} {n:#x}")  # d D 0xd
print(f"{big:,.1f}")        # 1,234,567.9 -- thousands separator
print(f"{0.256:.1%}")       # 25.6%
print(f"{'left':<10}|{'right':>10}|{'mid':^10}|")   # alignment
```

### Building tables

```
rows = [("Ada", 91.5), ("Grace", 88.25), ("Linus", 79.0)]
print(f"{'Name':<10}{'Score':>8}")
print("-" * 18)
for name, score in rows:
    print(f"{name:<10}{score:>8.1f}")
```

## 7.4 Regular expressions

The `re` module matches and transforms text using patterns. Always write patterns as **raw strings**.

| Pattern | Meaning |
|---|---|
| `\d` `\w` `\s` | digit, word character, whitespace |
| `.` | any character except newline |
| `+` `*` `?` | one or more, zero or more, optional |
| `{n,m}` | between *n* and *m* repetitions |
| `^` `$` | start / end of string |
| `( )` | capture group |
| `[abc]` | any one of the listed characters |

```
import re

print(re.findall(r"q\d+", "q0 q1 q10"))              # ['q0', 'q1', 'q10']

print(bool(re.match(r"q\d+", "q10_data")))           # True  -- match anchors at the START
print(bool(re.match(r"\d+", "shot_1024")))           # False

m = re.search(r"\d+", "shot_1024")                    # search scans the whole string
print(m.group())                                      # '1024'

print(re.sub(r"\d+", "#", "q0_measured_2"))           # q#_measured_#
```

Capture groups, named groups, and compiled patterns:

```
import re

log = "2026-10-02 ERROR disk full"
m = re.search(r"(\d{4})-(\d{2})-(\d{2}) (\w+)", log)
print(m.groups())                    # ('2026', '10', '02', 'ERROR')
print(m.group(4))                    # ERROR

pattern = re.compile(r"(?P<user>\w+)@(?P<domain>[\w.]+)")
m = pattern.search("contact: sam@example.com")
print(m.group("user"), m.group("domain"))

print(re.split(r"[,;]\s*", "a, b;c,  d"))
```

`match()` returns `None` when nothing matches, so check the result before calling `.group()`.

## Common pitfalls

- Expecting `str` methods to modify the string in place.
- Building a long string with `+=` inside a loop; collect pieces in a list and `"".join(...)` instead.
- Using `match()` when you meant `search()`.
- Writing regex patterns without the `r` prefix and fighting backslashes.
