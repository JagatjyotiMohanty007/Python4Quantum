## 6.1 Choosing a collection

| Type | Ordered | Mutable | Duplicates | Typical use |
|---|---|---|---|---|
| `list` | yes | yes | yes | sequence that changes |
| `tuple` | yes | no | yes | fixed record, e.g. a coordinate |
| `dict` | yes (insertion order) | yes | unique keys | lookup table, configuration, counts |
| `set` | no | yes | no | uniqueness, membership, set algebra |

Lists, tuples, and strings are **sequence types** (ordered, indexable). Dictionaries are the core **mapping type**. Sets are **set-like** types that support union, intersection, and difference.

## 6.2 Lists

```
numbers = [0, 1, 2]
numbers.append(3)
print(numbers)                 # [0, 1, 2, 3]

items = [3, 1, 2]
items.sort()                   # in place
print(items)                   # [1, 2, 3]
print(items.pop())             # 3 -- removes and returns the last item
items.reverse()
print(items)

nums = [3, 1, 2]
nums.append(4)
nums.remove(1)                 # removes the first matching VALUE
nums.insert(0, 99)
nums.extend([7, 8])
print(nums, nums.index(2), nums.count(7), len(nums))
```

| Method | Effect |
|---|---|
| `append(x)` | add to the end |
| `extend(iterable)` | add several items |
| `insert(i, x)` | insert at position |
| `remove(x)` / `pop(i)` | delete by value / by index |
| `sort()` / `reverse()` | in-place reorder |
| `index(x)` / `count(x)` | search |

Most mutating list methods return `None` — `result = items.sort()` leaves `result` as `None`.

## 6.3 Tuples

Tuples are immutable sequences, ideal for fixed groups of related values.

```
point = (1, 2)
print(point[0], point[1])

pair = (0, 1)                  # e.g. (source, destination)
single = (5,)                  # the comma makes the tuple, not the parentheses
print(type(single), type((5)))

lst = [1, 2]
tpl = (1, 2)
lst.append(3)
print(lst)
# tpl.append(3)   # AttributeError: tuples cannot change
# tpl[0] = 9      # TypeError
```

Choose a tuple when the data is a fixed record; choose a list when it grows or changes. Tuples are also **hashable** (if their items are), so they can be dictionary keys.

## 6.4 Dictionaries

```
settings = {"retries": 3, "timeout": 30}
print(settings["retries"])

settings["debug"] = True            # add
settings["timeout"] = 60            # update
print(settings.get("missing"))               # None -- no KeyError
print(settings.get("missing", "default"))    # default value
print("timeout" in settings)                 # membership tests KEYS

counts = {"00": 512, "11": 488}
print(counts.keys(), counts.values(), counts.items())

settings.update({"retries": 5, "mode": "fast"})   # merge another dict in
print(settings)

merged = {"a": 1} | {"b": 2}        # Python 3.9+: new merged dict
print(merged)

removed = settings.pop("debug")
print(removed, settings)
counts.setdefault("01", 0)          # insert only if absent
print(counts)
```

Iterating a dictionary:

```
scores = {"ada": 90, "grace": 85}

for key in scores:                       # keys
    print(key)
for value in scores.values():
    print(value)
for key, value in scores.items():        # most common
    print(key, value)
```

## 6.5 Sets

```
bits = {0, 1, 1, 0}
print(bits)                      # {0, 1} -- duplicates removed

active = {0, 1, 2, 2, 3}
active.add(4)
active.discard(99)               # no error if absent (remove() would raise)
print(active)

a = {1, 2, 3}
b = {2, 3, 4}
print(a | b)    # union:        {1, 2, 3, 4}
print(a & b)    # intersection: {2, 3}
print(a - b)    # difference:   {1}
print(a ^ b)    # symmetric difference: {1, 4}
print({1, 2} <= a)   # subset test

print(2 in a)        # membership in a set is very fast
unique = set([3, 1, 3, 2, 1])
print(sorted(unique))
empty_set = set()    # {} would create an empty DICT
```

## 6.6 Indexing and slicing

Indexes start at `0`; negative indexes count from the end. A slice `seq[start:stop:step]` includes `start` and excludes `stop`, and never raises `IndexError`.

```
letters = ["a", "b", "c", "d", "e"]
print(letters[0], letters[-1])        # a e
print(letters[1:4])                   # ['b', 'c', 'd']
print(letters[:2], letters[3:])       # ['a', 'b'] ['d', 'e']
print(letters[::2])                   # ['a', 'c', 'e']
print(letters[::-1])                  # reversed copy
print(letters[10:20])                 # [] -- slices are forgiving

word = "quantum"
print(word[0], word[1:4], word[-3:])
```

## 6.7 Unpacking

```
a, b = 1, 2
a, b = b, a                       # swap without a temp variable
print(a, b)                       # 2 1

first, *rest = [10, 20, 30, 40]   # starred target collects the rest
print(first, rest)
head, *middle, tail = range(6)
print(head, middle, tail)         # 0 [1, 2, 3, 4] 5

(x, y), z = (1, 2), 3             # nested unpacking
print(x, y, z)

for name, score in {"ada": 90}.items():
    print(name, score)
```

## 6.8 Hashable objects

Dictionary keys and set members must be **hashable**: immutable objects whose hash never changes. Strings, numbers, and tuples of hashables qualify; lists, dicts, and sets do not.

```
grid = {(0, 0): "start", (3, 4): "goal"}   # tuple keys work
print(grid[(3, 4)])

try:
    bad = {[1, 2]: "list key"}
except TypeError as err:
    print("TypeError:", err)
```

## Common pitfalls

- Using `{}` and expecting an empty set.
- Assuming `list.sort()` returns the sorted list.
- Mutating a list while iterating over it.
- Forgetting that `in` on a dictionary tests keys, not values.
- `a = [[0] * 3] * 3` creates three references to the *same* inner list (see Module 10).
