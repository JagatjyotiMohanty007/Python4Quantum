## 5.1 `while` loops

A `while` loop repeats as long as its condition is true. Make sure something inside the loop changes the condition.

```
count = 0
while count < 3:
    print(count)
    count += 1
```

## 5.2 `for` loops and `range()`

A `for` loop walks through any **iterable**: lists, strings, dictionaries, ranges, files, and more.

```
for i in range(3):          # 0, 1, 2
    print(i)

for ch in "abc":
    print(ch)

print(list(range(5)))              # [0, 1, 2, 3, 4]
print(list(range(2, 8)))           # [2, 3, 4, 5, 6, 7]
print(list(range(0, 10, 2)))       # [0, 2, 4, 6, 8]
print(list(range(5, 0, -1)))       # [5, 4, 3, 2, 1]
```

`range(start, stop, step)` includes `start` and excludes `stop`.

## 5.3 `enumerate()` and `zip()`

Avoid manual index counters. `enumerate()` yields `(index, value)` pairs; `zip()` walks several iterables in lockstep and stops at the shortest.

```
names = ["ada", "grace", "linus"]

for idx, value in enumerate(names):
    print(idx, value)

for position, value in enumerate(names, start=1):
    print(f"{position}. {value}")

scores = [90, 85, 77]
for name, score in zip(names, scores):
    print(name, score)

# Adjacent pairs: zip a list with itself shifted by one
steps = [0, 1, 2, 3]
for current, following in zip(steps, steps[1:]):
    print(f"{current} -> {following}")

# Build a dict from two lists
print(dict(zip(names, scores)))
```

## 5.4 Controlling loops: `break`, `continue`, `pass`

```
for i in range(5):
    if i == 3:
        break               # leave the loop immediately
    print(i)                # 0 1 2

for i in range(5):
    if i == 2:
        continue            # skip the rest of this iteration
    print(i)                # 0 1 3 4

for i in range(2):
    pass                    # placeholder: syntactically needs a body
print("done")
```

## 5.5 The loop `else` clause

An `else` attached to a loop runs **only if the loop finished without hitting `break`**. It is perfect for "search and report not found".

```
for i in range(3):
    print(i)
else:
    print("finished normally")

readings = ["001", "010", "110", "111", "000"]
target = "111"

for item in readings:
    if item == "000":
        continue            # skip a value we do not care about
    if item == target:
        print(f"Found {target}! Stopping search.")
        break
else:
    print(f"{target} not found in the data.")
```

## 5.6 Infinite loops, on purpose and by accident

`while True:` is a legitimate pattern when the exit lives in a `break`. An *accidental* infinite loop forgets to update the condition.

```
# Intentional: loop until the user supplies valid input
attempts = iter(["abc", "", "42"])     # stands in for repeated input() calls
while True:
    raw = next(attempts)
    if raw.isdigit():
        number = int(raw)
        break
    print(f"{raw!r} is not a number, try again")
print("got", number)
```

> In notebooks, an accidental infinite loop means *interrupt the kernel*. In a terminal, press `Ctrl+C`.

## 5.7 Convergence loops

Many algorithms repeat until a result is "good enough". Always include **both** a tolerance test and a maximum iteration count, so the loop cannot run forever.

```
# Newton's method for the square root of 2
target = 2.0
guess = 1.0
tolerance = 1e-10
max_iterations = 50
iteration = 0

while abs(guess * guess - target) > tolerance and iteration < max_iterations:
    guess = (guess + target / guess) / 2
    iteration += 1
    print(f"iteration {iteration:02d}: {guess:.10f}")

print(f"converged to {guess:.10f} in {iteration} iterations")
```

## 5.8 Nested loops

Loops can be nested; `break` exits only the innermost loop.

```
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end="\t")
    print()                 # newline after each row
```

## Common pitfalls

- Modifying a list while looping over it (build a new list or loop over a copy).
- Off-by-one errors with `range` (the stop value is excluded).
- A `while` condition that never becomes false.
- Using `range(len(x))` when `enumerate(x)` is clearer.
